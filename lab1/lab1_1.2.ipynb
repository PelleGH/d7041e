```python

import numpy as np

data = np.load("vecs.npy", allow_pickle=True).item()

print(type(data))
print(data.keys())
print(data["1_pos"].keys())
```

```python

def out(pos): 
    X = [] # embeddings
    y = [] # digits

    position = data[f"{pos}_pos"]

    for digit, embeddings in position.items():
        for embedding in embeddings:
            X.append(embedding)
            y.append(digit)

    X = np.array(X)
    y = np.array(y)

    return X, y
```

```python
X, y = out(1)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("First 10 labels:", y[:10])
print("Last labels:    ", y[len(y)-10:])
```

```python
rng = np.random.default_rng(67)

# random indices
indices = rng.permutation(len(X)) 

X_shuffled = X[indices]
y_shuffled = y[indices]
```

```python
print("Original labels:", y[:20])
print("Shuffled labels:", y_shuffled[:20])

# Check every shuffled sample against its original pair
pairs_correct = all(
    np.array_equal(X_shuffled[i], X[indices[i]])
    and y_shuffled[i] == y[indices[i]]
    for i in range(len(X))
)

print("All pairs preserved:", pairs_correct)
```
